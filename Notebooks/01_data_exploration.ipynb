```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
plt.figure(figsize=(10,6))

sns.countplot(
    data=df,
    x="Contract",
    hue="Customer Status"
)

plt.title("Customer Status by Contract Type", fontsize=16)
plt.xlabel("Contract Type", fontsize=12)
plt.ylabel("Number of Customers", fontsize=12)

plt.grid(axis="y", linestyle="--", alpha=0.3)

plt.tight_layout()

plt.show()
```

### Business Insight

Customers with month-to-month contracts have the highest churn rate.

Customers with one-year and two-year contracts are significantly more likely to stay with the company.

This suggests that longer contract commitments improve customer retention.

### Business Recommendation

The company should encourage customers to switch from month-to-month contracts to one-year or two-year contracts.

Possible strategies include:

- Loyalty discounts
- Free upgrades
- Bundle offers
- Exclusive long-term benefits

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    data=df,
    x="Customer Status",
    y="Monthly Charge"
)

plt.title("Monthly Charge by Customer Status", fontsize=16)
plt.xlabel("Customer Status")
plt.ylabel("Monthly Charge")

plt.grid(axis="y", linestyle="--", alpha=0.3)

plt.tight_layout()

plt.show()
```

### Business Insight

Customers who churn tend to have higher monthly charges than customers who stay. The median monthly charge for churned customers is noticeably higher, suggesting that pricing may influence customer retention.

### Business Recommendation

The company should closely monitor customers with high monthly charges and consider targeted retention strategies such as loyalty discounts, premium benefits, or personalized offers to reduce churn.

### Business Recommendation

The analysis suggests that customers paying higher monthly charges are more likely to churn. Instead of offering discounts to all customers, the company should target high-risk customers with retention incentives.

Possible retention strategies include:

- Loyalty discounts after contract renewal
- Free ad-free streaming subscriptions
- Additional mobile data or premium benefits
- Exclusive long-term customer rewards

These incentives should be offered when customers renew or extend their contracts, encouraging long-term retention while minimizing unnecessary revenue loss.

```python
plt.figure(figsize=(10,6))

sns.histplot(
    data=df,
    x="Age",
    bins=15,
    kde=True
)

plt.title("Distribution of Customer Age", fontsize=16)
plt.xlabel("Age")
plt.ylabel("Number of Customers")

plt.grid(axis="y", linestyle="--", alpha=0.3)

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    data=df,
    x="Customer Status",
    y="Age"
)

plt.title("Age Distribution by Customer Status", fontsize=16)
plt.xlabel("Customer Status")
plt.ylabel("Age")

plt.grid(axis="y", linestyle="--", alpha=0.3)

plt.tight_layout()

plt.show()
```

### Business Insight

The median age of churned customers is slightly higher than that of customers who stayed or recently joined. This suggests that age alone may not be a strong indicator of churn and should be analyzed together with other factors such as contract type, tenure, and monthly charges.

### Business Recommendation

The company should not base retention strategies on age alone. Instead, age should be combined with other customer attributes to identify customers who are at a higher risk of churning and provide them with personalized retention offers.
